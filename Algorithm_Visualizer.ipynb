Algorithm Visualizer

Project Overview

Algorithm Visualizer is a Python project that demonstrates how the Bubble Sort algorithm works through step-by-step visualizations.

The project also analyzes the algorithm's performance by counting comparisons and swaps, and tests its behavior using different input cases.

Objectives

Understand how Bubble Sort orders a list of numbers.

Visualize comparisons and swaps using Python and Matplotlib.

Measure basic algorithm operations.

Test the algorithm with different input cases.

```python
numbers = [5, 2, 8, 1, 4]

print("Before sorting:", numbers)

for i in range(len(numbers) - 1):
    if numbers[i] > numbers[i + 1]:
        numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

    print("Current list:", numbers)

print("After one pass:", numbers)
```

```python
def bubble_sort(numbers):
    numbers = numbers.copy()

    for pass_number in range(len(numbers) - 1):
        for i in range(len(numbers) - 1 - pass_number):
            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

    return numbers


assert bubble_sort([5, 2, 8, 1, 4]) == [1, 2, 4, 5, 8]
assert bubble_sort([1, 2, 3]) == [1, 2, 3]
assert bubble_sort([3, 2, 1]) == [1, 2, 3]
assert bubble_sort([]) == []
assert bubble_sort([7]) == [7]
assert bubble_sort([4, 4, 2]) == [2, 4, 4]

print("All tests passed!")
```

```python
def bubble_sort_steps(numbers):
    numbers = numbers.copy()
    steps = [numbers.copy()]

    for pass_number in range(len(numbers) - 1):
        for i in range(len(numbers) - 1 - pass_number):
            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]
                steps.append(numbers.copy())

    return steps


numbers = [5, 2, 8, 1, 4]
steps = bubble_sort_steps(numbers)

print("Original list:", numbers)
print("Number of recorded steps:", len(steps))

for step_number, step in enumerate(steps):
    print("Step", step_number, ":", step)
```

```python
import matplotlib.pyplot as plt

for step_number, step in enumerate(steps):
    plt.figure(figsize=(6, 3))

    plt.bar(range(len(step)), step)

    plt.title("Bubble Sort - Step " + str(step_number))
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.xticks(range(len(step)))

    plt.show()
```

```python
import matplotlib.pyplot as plt

def visualize_bubble_sort(numbers):
    numbers = numbers.copy()

    for pass_number in range(len(numbers) - 1):
        for i in range(len(numbers) - 1 - pass_number):

            colors = ["skyblue"] * len(numbers)
            colors[i] = "orange"
            colors[i + 1] = "orange"

            plt.figure(figsize=(6, 3))
            plt.bar(range(len(numbers)), numbers, color=colors)

            plt.title("Comparing " + str(numbers[i]) + " and " + str(numbers[i + 1]))
            plt.xlabel("Index")
            plt.ylabel("Value")
            plt.xticks(range(len(numbers)))

            plt.show()

            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

    print("Sorted list:", numbers)


visualize_bubble_sort([5, 2, 8, 1, 4])
```

```python
def bubble_sort_analysis(numbers):
    numbers = numbers.copy()
    comparisons = 0
    swaps = 0

    for pass_number in range(len(numbers) - 1):
        for i in range(len(numbers) - 1 - pass_number):
            comparisons += 1

            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]
                swaps += 1

    return numbers, comparisons, swaps


numbers = [5, 2, 8, 1, 4]

sorted_numbers, comparisons, swaps = bubble_sort_analysis(numbers)

print("Original list:", numbers)
```

```python
print("Test started")

sorted_numbers, comparisons, swaps = bubble_sort_analysis([5, 2, 8, 1, 4])

print("Sorted list:", sorted_numbers)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
print("Test finished")
```

```python
test_cases = [
    [],
    [7],
    [1, 2, 3],
    [3, 2, 1],
    [4, 4, 2],
    [10, -1, 5, 0]
]

for test in test_cases:
    sorted_numbers, comparisons, swaps = bubble_sort_analysis(test)

    assert sorted_numbers == sorted(test), f"Test failed: {test}"

print("All test cases passed!")
```

```python
def merge_sort(numbers):
    if len(numbers) <= 1:
        return numbers.copy()

    middle = len(numbers) // 2

    left_half = merge_sort(numbers[:middle])
    right_half = merge_sort(numbers[middle:])

    return merge(left_half, right_half)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


numbers = [5, 2, 8, 1, 4]

print("Original list:", numbers)
print("Merge Sort result:", merge_sort(numbers))
```

```python
def merge_sort(numbers):
    if len(numbers) <= 1:
        return numbers.copy()

    middle = len(numbers) // 2
    left_half = merge_sort(numbers[:middle])
    right_half = merge_sort(numbers[middle:])

    return merge(left_half, right_half)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


numbers = [5, 2, 8, 1, 4]

print("Original list:", numbers)
print("Merge Sort result:", merge_sort(numbers))
```

```python
import time

numbers = [64, 25, 12, 22, 11, 90, 34, 7, 18, 50]

start = time.perf_counter()
bubble_result = bubble_sort(numbers)
bubble_time = time.perf_counter() - start

start = time.perf_counter()
merge_result = merge_sort(numbers)
merge_time = time.perf_counter() - start

print("Bubble Sort result:", bubble_result)
print("Merge Sort result:", merge_result)
print("Bubble Sort time:", bubble_time)
print("Merge Sort time:", merge_time)
print("Results match:", bubble_result == merge_result)
```

```python
import random
import time

random.seed(42)
numbers = random.sample(range(10000), 1000)

start = time.perf_counter()
bubble_result = bubble_sort(numbers)
bubble_time = time.perf_counter() - start

start = time.perf_counter()
merge_result = merge_sort(numbers)
merge_time = time.perf_counter() - start

print("Number of items:", len(numbers))
print("Results match:", bubble_result == merge_result)
print("Bubble Sort time:", round(bubble_time, 6), "seconds")
print("Merge Sort time:", round(merge_time, 6), "seconds")
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Algorithm": ["Bubble Sort", "Merge Sort"],
    "Input Size": [len(numbers), len(numbers)],
    "Time (seconds)": [bubble_time, merge_time]
})

comparison["Time (seconds)"] = comparison["Time (seconds)"].round(6)

comparison
```

```python
import matplotlib.pyplot as plt

algorithms = comparison["Algorithm"]
times = comparison["Time (seconds)"]

plt.figure(figsize=(7, 4))
plt.bar(algorithms, times)
plt.title("Bubble Sort vs Merge Sort")
plt.xlabel("Algorithm")
plt.ylabel("Execution Time (seconds)")
plt.tight_layout()
plt.show()
```
